### Bước 3: Tạo Đường Dẫn Cloudflare Tunnel & Kết Nối Với Tune-Text

```python
import re
import subprocess
import time

# Chạy Cloudflare Tunnel trỏ vào cổng Ollama (11434)
tunnel_process = subprocess.Popen(
    ["cloudflared", "tunnel", "--url", "http://localhost:11434"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    universal_newlines=True
)

print("🔍 Đang thiết lập đường truyền bảo mật Tunnel... Vui lòng đợi vài giây...")
public_url = None

for line in tunnel_process.stdout:
    match = re.search(r'https://[a-zA-Z0-9-]+\.trycloudflare\.com', line)
    if match:
        public_url = match.group(0)
        break

if public_url:
    print("\n===================================================================")
    print("🎉 ĐÃ KHỞI TẠO CLOUD GPU WORKER THÀNH CÔNG!")
    print("===================================================================")
    print(f"👉 ĐƯỜNG DẪN KẾT NỐI CỦA BẠN: {public_url}")
    print("===================================================================")
    print("📋 Hãy copy đường link trên và dán vào giao diện Tune-Text (mục Cloud GPU).")
    print("Hệ thống sẽ tự động gánh toàn bộ tải nặng trên GPU Google miễn phí!")
else:
    print("⚠️ Không tìm thấy URL Cloudflare Tunnel. Hãy kiểm tra lại kết nối mạng.")
```
